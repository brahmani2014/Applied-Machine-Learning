# Chapter 5 — Multiclass Logistic Regression

This notebook corresponds to Week 5 of *Applied Machine Learning* and uses the Forest Fires data. Place the book's `Data-Week5.csv` file in `data/raw/` at the repository root.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import ConfusionMatrixDisplay, accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split

data_path = Path('../../data/raw/Data-Week5.csv')
data = pd.read_csv(data_path)
data.head(3)
```

## Define and split the data


```python
x = data[['DMC', 'wind']]
y = data['month']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=42
)
```

## Train the model

The explicit `multi_class='multinomial'` argument is retained from the book. Recent scikit-learn versions may issue a deprecation warning because multinomial handling is becoming automatic.


```python
model = LogisticRegression(
    max_iter=200, multi_class='multinomial', solver='lbfgs'
)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

## Confusion matrix


```python
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)
display = ConfusionMatrixDisplay(
    confusion_matrix=cm, display_labels=model.classes_
)
display.plot(cmap='Blues', xticks_rotation=45)
plt.title('Logistic Regression Confusion Matrix')
plt.show()
```

## Accuracy


```python
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
```

Book result: accuracy `0.57`. The book concludes that logistic regression is not effective for this prediction task.
