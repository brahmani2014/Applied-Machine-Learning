# Chapter 6 — Decision Trees

This notebook corresponds to Week 6 of *Applied Machine Learning* and uses the Social Media vs Productivity dataset. Place the book's `Data-Week6.csv` file in `data/raw/` at the repository root.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree

data_path = Path('../../data/raw/Data-Week6.csv')
data = pd.read_csv(data_path)
data.head()
```

## Encode the social-platform feature

The assignment form below produces the same values as the book while avoiding deprecated pandas chained assignment.


```python
data['social_platform_preference'] = data['social_platform_preference'].replace(
    ['Facebook', 'Instagram', 'Telegram', 'TikTok', 'Twitter'],
    [0, 1, 2, 3, 4],
)
```

## Train and evaluate the full-data example


```python
x = data[['social_platform_preference', 'sleep_hours']]
y = data['stress_level']

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.25)
classifier = DecisionTreeClassifier()
classifier.fit(X_train, y_train)

print(accuracy_score(y_true=y_train, y_pred=classifier.predict(X_train)))
print(accuracy_score(y_true=y_test, y_pred=classifier.predict(X_test)))
```

Book results for one run: training accuracy `0.9109719058466211` and test accuracy `0.10293280182232346`.

## Plot a tree using the smaller subset


```python
x_subset = x.iloc[0:2745, :]
y_subset = y.iloc[0:2745]

X_train, X_test, y_train, y_test = train_test_split(
    x_subset, y_subset, test_size=0.25
)
classifier = DecisionTreeClassifier()
classifier.fit(X_train, y_train)

print(accuracy_score(y_true=y_train, y_pred=classifier.predict(X_train)))
print(accuracy_score(y_true=y_test, y_pred=classifier.predict(X_test)))

plt.figure(figsize=(20, 10))
plot_tree(classifier, filled=True)
plt.show()
```

Book results for the subset example: training accuracy `1.0` and test accuracy `0.9985443959243085`. The book relates social-platform preference and sleep hours to stress level.

The original example does not set random seeds, so both the split and tree can change between runs. The large difference between the two displayed test accuracies is preserved for review.
