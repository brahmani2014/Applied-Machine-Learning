# Chapter 2 — Loading and Visualizing Data

This notebook corresponds to Week 2 of *Applied Machine Learning*. It demonstrates loading the Iris dataset, creating common exploratory plots, reading a CSV file, and drawing a bar chart.


## Iris dataset


```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

## Histograms


```python
df.hist()
plt.show()
```

## Density plots


```python
df.plot(kind='density', subplots=True, layout=(2, 2), sharex=False)
plt.show()
```

## Box plots


```python
df.plot(kind='box', subplots=True, layout=(2, 2), sharex=False, sharey=False)
plt.show()
```

## Correlation heat map


```python
names = iris.feature_names
correlations = df.corr()
fig = plt.figure()
ax = fig.add_subplot(111)
cax = ax.matshow(correlations, vmin=-1, vmax=1)
fig.colorbar(cax)
ticks = np.arange(0, 4, 1)
ax.set_xticks(ticks)
ax.set_yticks(ticks)
ax.set_xticklabels(names)
ax.set_yticklabels(names)
plt.show()
```

## Scatter-plot matrix


```python
pd.plotting.scatter_matrix(df)
plt.show()
```
