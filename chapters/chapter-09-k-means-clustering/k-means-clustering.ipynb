# Chapter 9 — K-Means Clustering

This notebook corresponds to Week 9 of *Applied Machine Learning* and uses worldwide temperature data from 1995–2020. Place the book's `Data-Week9.csv` file in `data/raw/` at the repository root. The book notes that invalid rows were removed during preprocessing.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans

data_path = Path('../../data/raw/Data-Week9.csv')
data = pd.read_csv(data_path)
data.head()
```

## Elbow graph


```python
x = data[['Year', 'AvgTemperature']]
SSE = []
for k in range(1, 10):
    model = KMeans(n_clusters=k)
    model.fit(x)
    data['clusters'] = model.labels_
    SSE.append(model.inertia_)

plt.figure(figsize=(20, 10))
plt.plot(
    range(1, 10), SSE, linestyle='dashed', marker='o',
    markerfacecolor='r', markersize=10
)
plt.show()
```

## Four-cluster temperature model

Transcription correction: the PDF contains `' AvgTemperature'` with a leading space in this section, while the preceding code uses `AvgTemperature`. The consistent name is used below to prevent an immediate `KeyError`; the clustering method is unchanged.


```python
filtered_data = data[data['AvgTemperature'] > 0]
X1 = filtered_data[['Year']].values
X2 = filtered_data[['AvgTemperature']].values

model = KMeans(n_clusters=4)
model.fit(X2)
prediction = model.predict(X2)
plt.scatter(X1, X2, c=prediction, s=50, cmap='viridis')
centers = model.cluster_centers_
```

The original examples do not fix K-means initialization, so assignments and inertia values can vary between runs and library versions.
