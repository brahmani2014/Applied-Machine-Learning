# Chapter 7 — K-Nearest Neighbors

This notebook corresponds to Week 7 of *Applied Machine Learning* and uses the Depression and Anxiety dataset. Place the book's `Data-Week7.csv` file in `data/raw/` at the repository root.

> This health-related example is for education only and is not intended for diagnosis or decisions about individuals.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import ListedColormap
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

data_path = Path('../../data/raw/Data-Week7.csv')
data = pd.read_csv(data_path)
data.head()
```

## Elbow graph


```python
x = data[['age', 'bmi']]
y = data['depressiveness']

SSE = []
for k in range(1, 20):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x, y)
    pred_i = knn.predict(x)
    SSE.append(np.mean(pred_i != y))

plt.figure(figsize=(20, 10))
plt.plot(
    range(1, 20), SSE, color='b', linestyle='dashed', marker='o',
    markerfacecolor='r', markersize=10
)
plt.gca().invert_yaxis()
plt.show()
```

## Six-neighbor classifier and decision regions


```python
n_neighbors = 6
X = x.values[:, 0:2]
h = 2
classifier = KNeighborsClassifier(n_neighbors, weights='distance')
classifier.fit(X, y)

x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(
    np.arange(x_min, x_max, h), np.arange(y_min, y_max, h)
)
cmap_light = ListedColormap(['lightgreen', 'lightblue', 'yellow'])
cmap_dark = ListedColormap(['blue', 'green', 'red'])
Z = classifier.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure()
plt.pcolormesh(xx, yy, Z, cmap=cmap_light)
plt.scatter(X[:, 0], X[:, 1], c=y, cmap=cmap_dark)
plt.xlim(xx.min(), xx.max())
plt.ylim(yy.min(), yy.max())
plt.title('6-NN')
plt.show()
```

## Accuracy shown in the book

> Methodology note: the following sequence is preserved from the book. The classifier was fitted on all observations before the split, so the reported test score is not a held-out evaluation. Correcting this would change the method and potentially the results.


```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
train_accuracy = classifier.score(X_train, y_train)
test_accuracy = classifier.score(X_test, y_test)
print(train_accuracy)
print(test_accuracy)
```

Book results: training accuracy `0.9630818619582665` and test accuracy `0.9551282051282052`.
