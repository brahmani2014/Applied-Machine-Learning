# Chapter 10 — Hierarchical Clustering

This notebook corresponds to Week 10 of *Applied Machine Learning* and uses the NYC Air Quality and Health Impacts dataset. Place the book's `Data-Week10.csv` file in `data/raw/` at the repository root.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import dendrogram, linkage

data_path = Path('../../data/raw/Data-Week10.csv')
data = pd.read_csv(data_path)
data.head()
```

## Encode geography type and select observations

The assignment form below produces the same codes as the book while avoiding deprecated pandas chained assignment.


```python
data['Geo Type Name'] = data['Geo Type Name'].replace(
    ['Borough', 'CD', 'Citywide', 'UHF34', 'UHF42'],
    [0, 1, 2, 3, 4],
)
df = data[['Geo Type Name', 'Data Value']]
df = df.iloc[:1000, :]
```

## Ward-linkage dendrogram


```python
linked = linkage(df, method='ward')

plt.figure(figsize=(8, 5))
dendrogram(
    linked,
    orientation='top',
    distance_sort='descending',
    show_leaf_counts=True,
)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.show()
```
